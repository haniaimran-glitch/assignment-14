Dataset 1: Pakistani E-Commerce Data

```python
pak_ecom=pd.read_csv('pakistani_ecommerce_dataset.csv')
```

Q1: Which is the most popular product category? 

```python
pop_cat=pak_ecom['product_category'].value_counts()
print(pop_cat)
if pop_cat['Home Decor'] > pop_cat['Clothing'] and pop_cat['Home Decor'] > pop_cat['Beauty'] and pop_cat['Home Decor'] > pop_cat['Electronics'] and pop_cat['Home Decor'] > pop_cat['Books']:
    print("Home Decor is the most popular category")
elif pop_cat['Clothing'] > pop_cat['Home Decor'] and pop_cat['Clothing'] > pop_cat['Beauty'] and pop_cat['Clothing'] > pop_cat['Electronics'] and pop_cat['Clothing'] > pop_cat['Books']:
    print("Clothing is the most popular category")
elif pop_cat['Beauty'] > pop_cat['Home Decor'] and pop_cat['Beauty'] > pop_cat['Clothing'] and pop_cat['Beauty'] > pop_cat['Electronics'] and pop_cat['Beauty'] > pop_cat['Books']:
    print("Beauty is the most popular category")
elif pop_cat['Electronics'] > pop_cat['Home Decor'] and pop_cat['Electronics'] > pop_cat['Clothing'] and pop_cat['Electronics'] > pop_cat['Beauty'] and pop_cat['Electronics'] > pop_cat['Books']:
    print("Electronics is the most popular category")
else:
    print("Books is the most popular category")
```

Q2: What is the average order value?

```python
#calculate the average order value and round it to 2 decimal places
avg=pak_ecom['order_value'].mean().round(2)
print('The average order value is',avg)
```

Q3: Which city has the highest number of orders? 

```python
city_orders=pak_ecom['city'].value_counts()
print(city_orders)
if city_orders['Karachi'] > city_orders['Lahore'] and city_orders['Karachi'] > city_orders['Islamabad'] and city_orders['Karachi'] > city_orders['Rawalpindi'] and city_orders['Karachi'] > city_orders['Faisalabad']:
    print("Karachi has the highest number of orders")
elif city_orders['Lahore'] > city_orders['Karachi'] and city_orders['Lahore'] > city_orders['Islamabad']and city_orders['Lahore'] > city_orders['Rawalpindi']and city_orders['Lahore'] > city_orders['Faisalabad']:
    print("Lahore has the highest number of orders")
elif city_orders['Islamabad'] > city_orders['Karachi'] and city_orders['Islamabad'] > city_orders['Lahore']and city_orders['Islamabad'] > city_orders['Rawalpindi']and city_orders['Islamabad'] > city_orders['Faisalabad']:
    print("Islamabad has the highest number of orders")
else:
    print("Faisalabad has the highest number of orders")
```

Q4: What is the most used payment method? 

```python
pop_pay=pak_ecom['payment_method'].value_counts()
print(pop_pay)
if pop_pay['Cash on Delivery'] > pop_pay['JazzCash'] and pop_pay['Cash on Delivery'] > pop_pay['EasyPaisa'] and pop_pay['Cash on Delivery'] > pop_pay['Debit Card']:
    print("Cash on Delivery is the most used payment method")
elif pop_pay['JazzCash'] > pop_pay['Cash on Delivery'] and pop_pay['JazzCash'] > pop_pay['EasyPaisa'] and pop_pay['JazzCash'] > pop_pay['Debit Card']:
    print("JazzCash is the most used payment method")
elif pop_pay['EasyPaisa'] > pop_pay['Cash on Delivery'] and pop_pay['EasyPaisa'] > pop_pay['JazzCash'] and pop_pay['EasyPaisa'] > pop_pay['Debit Card']:
    print("EasyPaisa is the most used payment method")
else:
    print("Debit Card is the most used payment method")
```

Q5: Draw a bar chart showing average order value by city

```python
import matplotlib.pyplot as plt
```

```python
#calculate the average order value for each city
avg_city = pak_ecom.groupby('city')['order_value'].mean()
```

```python
#put city names in categories variable
categories = avg_city.index
#put average order values in values variable
values = avg_city.values
#draw the bar chart
plt.bar(categories, values, color='violet', width=0.5)
plt.xlabel('City')
plt.ylabel('Average Order Value')
plt.title('Average Order Value by City')
plt.show()
```
